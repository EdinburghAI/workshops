# What's this workshop about? 🤔

A coach reviewing a match first needs to find the moments worth watching. [Judo Clipper](https://www.judoclipper.com/) is a real application that helps find likely throw attempts.

In judo, athletes try to throw each other onto their backs. They can also win through groundwork, but today we'll focus on throws. A **throw attempt** means someone tries to execute a throw, whether or not the opponent ends up on the ground.

Today we'll explore how a computer could learn to recognise those attempts. We'll watch videos, inspect their individual pictures, turn body positions into numbers and invent our own detection rules.

You won't need to write or run code. This notebook is your activity guide.

# Task 1: A Video Is a Sequence of Pictures 🎞️

Open **1 · Video is pictures** in the app.

A video is a sequence of still pictures shown quickly, one after another. Each picture is called a **frame**.

These clips have 30 frames every second. Seven seconds therefore gives us **210 frames**.

**Your Job**

1. Play the throw-attempt example.
2. Move the frame slider to inspect individual pictures.
3. Look at the pictures shown side by side in time order.
4. Switch to the no-attempt example and repeat.

**Questions**

- What can you tell from the sequence that you couldn't tell from a single picture?
- Could one bent-over pose happen both during a throw and during ordinary fighting?
- What would happen to the movement if we shuffled the pictures?

We want a model that can use how body positions **change over time**, rather than judging every frozen picture separately.

# Task 2: What Is a Picture to a Computer? 🔍

Open **2 · Pictures are numbers**.

Zoom far enough into a digital picture and you'll find a grid of tiny coloured squares. These are **pixels**.

A **bitmap** represents a picture as a grid of pixels, with numbers describing their colours. In the view we're using, each pixel has a red number, a green number and a blue number. Combining them describes its colour.

Computers store and process information using patterns of **ones and zeros**. Pictures are stored this way too: the numbers describing each pixel's colour are represented using ones and zeros, so the computer can work with them. This way of representing numbers is called **binary**. You don't need to learn how to calculate with it today.

**Your Job**

1. Use the position sliders to select a pixel on an athlete's judogi.
2. Inspect its red, green and blue values, and the binary digits underneath.
3. Select a pixel on the mat. How do the values change?
4. Leave the pixel position fixed and move through the video frames. What happens when an athlete moves through that position?

**More movement doesn't mean more ones.** Movement changes the values and arrangement of the pixels. The challenge is learning which changing patterns are relevant to our question.

Video files usually compress the information to save space. The decoded frames let us inspect those pictures again.

# Task 3: Which Information Actually Matters? 🦴

Open **3 · Players become skeletons**.

Our original pictures contain the athletes, the referee, advertising boards and the crowd. We want to find throws, so the athletes' movement is particularly useful.

A pose-estimation model estimates positions of joints such as shoulders, elbows, hips and knees. Each position becomes two numbers: how far **across** the picture it is, and how far **down** it is.

This gives us a skeleton representation. We aren't manually drawing the correct answer into the image; these are an AI model's estimates, and they can be wrong.

**Your Job**

1. Switch between **Original picture**, **Skeleton overlay** and **Skeleton only**.
2. Highlight a joint and move through the frames.
3. Inspect how its two position numbers change.
4. Open **Watch this skeleton sequence move** to see those positions as an animation.

**Questions**

- Can you still recognise the action with the background removed?
- Which useful information did we lose, such as details of someone's grip?
- Do any joints disappear or jump to the wrong person?

Using skeletons reduces the background information our throw classifier receives. It doesn't eliminate every problem: camera movement, framing and incorrect pose estimates still affect the numbers.

## A Video Becomes a Sequence of Lists 🔢

Let's connect the ideas:

- One athlete has 17 estimated joints.
- Each joint has two position numbers.
- Two athletes therefore give us **68 numbers for one frame**.
- Seven seconds gives us **210 frames**.

So the video becomes **210 ordered lists of 68 numbers**.

Just like the numbers describing pixel colours, these joint-position numbers are stored using **ones and zeros** so the computer can work with them. Our model uses the changing positions to learn patterns of movement that tend to appear in throw attempts.

# How Does the Real Model Learn? 🧠

You described a rule in words. The real model learns numerical relationships from many examples whose labels tell it whether a clip contains an attempt.

A **Recurrent Neural Network (RNN)** reads the frame representations in order and updates a running numerical summary as it goes. Earlier frames can therefore influence how it interprets later ones.

A **Long Short-Term Memory network (LSTM)** adds learned controls that help it retain, update and expose information in that summary.

After the sequence, a classifier uses the summary to make a prediction about the whole clip.

The model doesn't need to know the names of judo techniques. During training, its internal settings change to reduce errors on the labelled examples. We then evaluate it on separate data to check whether the patterns it learned are useful beyond those examples.

**Discuss:** why might seeing the setup help interpret what happens when the athletes overlap?

# Task 5: A Fair Final Test 🎯

Open **5 · Final quiz** only after you finish revising your rule on the practice clips.

**Your Job**

1. Write your final rule.
2. Watch the new examples.
3. Commit a prediction for every clip.
4. Submit all your answers, then inspect the results.

If you now change your rule using these answers and repeat the same quiz, your improved score wouldn't be a fair check on new examples: you already know what happened in them.

Engineers follow the same principle:

- **Training data** teaches the model.
- **Validation data** helps choose settings and compare versions.
- **Test data** checks the finished choice after those decisions are complete.

**Don't tune a model's settings using its test results.** That would let the final answers influence the model-development process and could make its performance look better than it really is.

This small activity is an illustration of that principle, not a measurement of the real model's accuracy. The app's personal quiz roles are separate from the research dataset's splits.
